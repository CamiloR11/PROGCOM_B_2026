#WHILE
1. Cuando no se sabe cuantas veces se rpeite una accion
2. Es gobernado por una condicion de parada
3. Es necesaria una condicion inicial, la cual **DEBE** cambiar durante el ciclo

Sintaxis
```
while(condicional)
- Codigo a repetir
- Modificacion de condicion
```
***OJO***
La modificacion de la condicion puede darse de diferentes formas:
1. contador
2. Un acumulador
3. Cambio de bandera (trabajo con bolenaos)
4. Con un imp

````python
#Suponer que la cantidad maxima de votantes son 10
#Suponer que son 5 candidatos
v1=0
v2=0
v3=0
v4=0
v5=0
vb=0
vn=0
cedulas=[]
votos=0
while votos<10:
    id=input("Ingrese su cc ")
    print("")21
    if id in cedulas:
        print("Esta cédula ya votó.\n")
        continue
    cedulas.append(id)
    voto=int(input("Ingrese por quien quiere votar: \n1.Elefante blanco\n2.Pantera rosa\n3.JP\n4.Paloma\n5.Quevedo\n6.voto en blanco."))
    if voto==1: v1+=1
    elif voto==2: v2+=1
    elif voto==3: v3+=1
    elif voto==4: v4+=1
    elif voto==5: v5+=1
    elif voto==6: vb+=1
    else: vn+=1
    votos+=1
    print("Gracias por su voto.\n")
    if votos==5:
        resp=input("¿Faltan votantes? (s/n): ")
        if resp=="s":
            votos=0
        else:
            break
print("\nRESULTADO")
print("Elefante blanco:",v1)
print("Pantera Rosa:",v2)
print("JP:",v3)
print("Paloma:",v4)
print("Quevedo:",v5)
print("Voto en blanco:",vb)
ganador=max(v1,v2,v3,v4,v5)
if ganador==v1:
    print("Ganador: Elefante blanco")
elif ganador==v2:
    print("Ganador: Pantera Rosa")
elif ganador==v3:
    print("Ganador: JP")
elif ganador==v4:
    print("Ganador: Paloma")
elif ganador==v5:
    print("Ganador: Quevedo")
````
