#Programación orientada a objetos POO
- Paradigma de programación enfocado en crear marcos (clase) que a su vez generan instancias (objetos) •
- Está fundamentada bajo 4 princípios fundamentales o pilares: **Herencia Abstracción, Polímorfismo y
Encapsulamiento**.
- Se enfoca en crear plantillas (class) las cuales desarrollan instancias(objetos). Esos objetos hacen cosas(metodos), y tienen caracteristicas propias(atributos).

" Ventajas al implementar Poo
1. Escalabilidad: un mismo código en diferentes app.
2. Seguridad: No cualquier persona tiene acceso al código.
3. Modularidad: El código se expresa a partir de
"pedazos" (métodos).
4. Reusabilidad: Se puede reutilizar partes del código
en lugar de generar un código nuevo.
5. Mantenibilidad: El código puede ser fácilmente
corregido y actualizado.
