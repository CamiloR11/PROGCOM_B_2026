```python
class Persona:
    def __init__(self, nombre, color_camiseta, color_pantalon, color_pelo):
        self.nombre = nombre
        self.color_camiseta = color_camiseta
        self.color_pantalon = color_pantalon
        self.color_pelo = color_pelo

    # Mostrar características
    def mostrar_caracteristicas(self):
        print(f"\n===== Características de {self.nombre} =====")
        print(f"Color de camiseta: {self.color_camiseta}")
        print(f"Color de pantalón: {self.color_pantalon}")
        print(f"Color de pelo:     {self.color_pelo}")

    # PASOS DEL BAILE
    def paso1(self):
        print("\nPASO 1")
        print(f"{self.nombre} toma posición inicial")

    def paso2(self):
        print("\nPASO 2")
        print(f"{self.nombre} levanta la pierna derecha")

    def paso3(self):
        print("\nPASO 3")
        print(f"{self.nombre} mantiene el equilibrio")

    def paso4(self):
        print("\nPASO 4")
        print(f"{self.nombre} baja el pie y da media vuelta")

    def paso5(self):
        print("\nPASO 5")
        print(f"{self.nombre} levanta el pie derecho y da una patada")

    def paso6(self):
        print("\nPASO 6")
        for i in range(2):
            print(f"{self.nombre} da mini saltos laterales")

    def paso7(self):
        print("\nPASO 7")
        print(f"{self.nombre} levanta el pie derecho, da patadas y avanza")

    def paso8(self):
        print("\nPASO 8")
        print(f"{self.nombre} levanta las manos y cruza un pie detrás del otro")

    def paso9(self):
        print("\nPASO 9")
        print(f"{self.nombre} mueve piernas hacia dentro y afuera en saltos")

    def paso10(self):
        print("\nPASO 10")
        print(f"{self.nombre} estira mano izquierda y pie derecho")

    def paso11(self):
        print("\nPASO 11")
        print(f"{self.nombre} mueve caderas de lado a lado")

    def paso12(self):
        print("\nPASO 12")
        print(f"{self.nombre} se hace el muerto 💀")
        print("Fin del baile 👏")


# ── Crear bailarines ─────────────────────────────
Camilo = Persona("Camilo", "Azul Oscuro", "Negro", "Negro")
Andres = Persona("Andres", "Negro", "Gris", "Negro")

# ── Mostrar características ──────────────────────
Camilo.mostrar_caracteristicas()
Andres.mostrar_caracteristicas()

# ── Lista de pasos ───────────────────────────────
pasos = [
    "paso1",
    "paso2",
    "paso3",
    "paso4",
    "paso5",
    "paso6",
    "paso7",
    "paso8",
    "paso9",
    "paso10",
    "paso11",
    "paso12"
]

# ── Baile sincronizado ───────────────────────────
for paso in pasos:
    getattr(Camilo, paso)()
    getattr(Andres, paso)()
```
