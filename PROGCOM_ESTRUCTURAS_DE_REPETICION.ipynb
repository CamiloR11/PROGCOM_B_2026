#ESTRUCTURAS DE REPETICION
1.  for. Cuando se sabe cuantas veces se debe repetir una accion
2. While. No se sabe cuantas veces se repite la accion

##1. FOR
Sintaxis general en phyton

for (variable) in (algo):
se repite una acccion

algo
* in range(n): quiero repetir n veces
* in range(CI,CF,p): Quiero repetir desde CI, hasta CF-1, con paso p
* in (lista, diccionario): Cuando quiero iterar cada elemento dentro de la lista/diccionario
* each (lista, diccionario): Cuando quiero iterar cada elemento dentro de la lista/diccionario


```python
#Quiero repetir n veces
for i in range(9):
  print(f"Hay {i+1} personas en el salon")
```

```python
for i in range (1,10):
  print(f"Hay {i} personas en el salon")
```

```python
dias=int(input("Ey!! Cuantos dias faltan para tu cumpleaños?"))
for dias in range(dias,0,-1):
  print(f"Faltan {dias} dias para tu cumpleaños!!")
```

#WHILE
Sintaxis
'''
while (VARIABLE=CI):
Lo que se va a repetir
Modificador de CI
'''
**OJO**
- While arranca si CI es VERDADERA
- Modificaciones de CI: PREGUNTA, CONTADOR, ACUMULADOR, CAMBIO DE VARIABLE


```python
c1=0
c2=0
c3=0
nulo=0
voto=int(input("Ingrese el numero de su candidato (1-3). \n0 para voto en blanco"))
if voto==1: c1=c1+1
elif voto==2: c2=c2+1
elif voto==3: c3=c3+1
else:nulo+1
print(f"total de votos:\nCandidato 1={c1}\nCandidato 2={c2}\nCandidato 3={c3}\nNulo={nulo}")
```
