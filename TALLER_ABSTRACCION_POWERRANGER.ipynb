```python
class CajeroAutomatico:

    # Método visible para el usuario

    def retirar_dinero(self, monto):

        print(f"Retirando ${monto}...")

        # Procesos internos ocultos

        self.__verificar_saldo(monto)

        self.__actualizar_base_datos(monto)

    # Método visible para el usuario
    def consultar_saldo(self):

        print("Consultando saldo...")

        # Procesos internos ocultos

        self.__conectar_servidor()

        self.__leer_cuenta()

    # Método visible para el usuario

    def transferir_dinero(self, monto, destino):

        print(f"Transfiriendo ${monto} a {destino}...")

        # Procesos internos ocultos

        self.__validar_destino(destino)

        self.__registrar_movimiento()


    # MÉTODOS PRIVADOS

    def __verificar_saldo(self, monto):

        print("Verificando saldo disponible")

    def __actualizar_base_datos(self, monto):

        print("Actualizando base de datos")

    def __conectar_servidor(self):

        print("Conectando con el servidor")

    def __leer_cuenta(self):

        print("Leyendo información de la cuenta")

    def __validar_destino(self, destino):

        print("Validando cuenta destino")

    def __registrar_movimiento(self):

        print("Registrando movimiento")
```

```python
class Personaje:

    def __init__(self, nombre, energia):
        self.nombre = nombre
        self.energia = energia

    # Método público
    def atacar(self, armadura_enemigo):

        daño = self.__calcular_daño(armadura_enemigo)

        print(f"{self.nombre} hizo {daño} de daño")

    # Método privado oculto
    def __calcular_daño(self, armadura_enemigo):

        daño_base = 50

        # Influencia de la energía
        if self.energia < 30:
            daño_base -= 15

        # Influencia de la armadura
        daño_final = daño_base - armadura_enemigo

        if daño_final < 0:
            daño_final = 0

        return daño_final
```

```python
class SistemaEnvios:

    # Método abstracto y genérico
    def enviar_paquete(self, destino, peso):

        print(f"Enviando paquete a {destino}")

        # El sistema decide el transporte automáticamente
        if peso < 2:
            self.__usar_drone()

        elif peso < 100:
            self.__usar_camion()

        else:
            self.__usar_barco()

    # Métodos privados ocultos
    def __usar_drone(self):
        print("Transporte: Drone")

    def __usar_camion(self):
        print("Transporte: Camión")

    def __usar_barco(self):
        print("Transporte: Barco")
```
