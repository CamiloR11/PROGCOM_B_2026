```python
#Ejercicio 1 - Clase Persona
class Persona:
    def __init__(self, edad):
        if edad > 0:
            self.__edad = edad
        else:
            self.__edad = 0

    def get_edad(self):
        return self.__edad

    def set_edad(self, nueva_edad):
        if nueva_edad > 0:
            self.__edad = nueva_edad
        else:
            print("Error: la edad debe ser mayor a 0.")
```

```python
#Ejercicio 2 - Clase Libro
class Libro:
    def __init__(self, titulo):
        self.__titulo = titulo

    def get_titulo(self):
        return self.__titulo

    def set_titulo(self, nuevo_titulo):
        if nuevo_titulo.strip():
            self.__titulo = nuevo_titulo
        else:
            print("Error: el título no puede estar vacío.")
```

```python
#Ejercicio 3 - Clase Cuentabancaria Saldo Negativo
class CuentaBancaria:
    def __init__(self, saldo_inicial=0):
        if saldo_inicial >= 0:
            self.__saldo = saldo_inicial
        else:
            self.__saldo = 0

    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
        else:
            print("Error: no se permiten depósitos negativos.")

    def get_saldo(self):
        return self.__saldo
```

```python
#Ejercicio 4 - Clase Cuentabancaria Retiro de dinero
class CuentaBancaria:
    def __init__(self, saldo_inicial=0):
        if saldo_inicial >= 0:
            self.__saldo = saldo_inicial
        else:
            self.__saldo = 0

    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
        else:
            print("Error: depósito inválido.")

    def retirar(self, cantidad):
        if cantidad <= 0:
            print("Error: el retiro debe ser positivo.")
        elif cantidad > self.__saldo:
            print("Error: saldo insuficiente.")
        else:
            self.__saldo -= cantidad

    def get_saldo(self):
        return self.__saldo
```

```python
#Ejercicio 5 - Clase Estudiante
class Estudiante:
    def __init__(self, nombre):
        self.nombre = nombre
        self.__nota = 0

    def set_nota(self, nota):
        if 0 <= nota <= 5:
            self.__nota = nota
        else:
            print("Error: la nota debe estar entre 0 y 5.")

    def mostrar_nota(self):
        print(f"Nota: {self.__nota}")
```

```python
#Ejercicio 6 - Clase Producto
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre

        if precio > 0:
            self.__precio = precio
        else:
            self.__precio = 0

    def get_precio(self):
        return self.__precio

    def set_precio(self, nuevo_precio):
        if nuevo_precio > 0:
            self.__precio = nuevo_precio
        else:
            print("Error: el precio debe ser positivo.")
```

```python
#Ejercicio 7 - Clase Temperatura
class Temperatura:
    def __init__(self, celsius=0):
        if celsius >= -273.15:
            self.__celsius = celsius
        else:
            self.__celsius = -273.15

    def get_celsius(self):
        return self.__celsius

    def set_celsius(self, nueva_temp):
        if nueva_temp >= -273.15:
            self.__celsius = nueva_temp
        else:
            print("Error: temperatura inválida.")
```

```python
#Ejercicio 8 - Clase Usuario
class Usuario:
    def __init__(self, nombre, password):
        self.nombre = nombre

        if len(password) >= 6:
            self.__password = password
        else:
            self.__password = "123456"

    def cambiar_password(self, nueva_password):
        if len(nueva_password) >= 6:
            self.__password = nueva_password
        else:
            print("Error: mínimo 6 caracteres.")

    def verificar_password(self, password):
        return self.__password == password
```

```python
#Ejercicio 9 - Clase Vehiculo
class Vehiculo:
    def __init__(self, marca):
        self.marca = marca
        self.__velocidad = 0

    def get_velocidad(self):
        return self.__velocidad

    def acelerar(self, incremento):
        if incremento >= 0:
            self.__velocidad += incremento
        else:
            print("Error: velocidad inválida.")
```
