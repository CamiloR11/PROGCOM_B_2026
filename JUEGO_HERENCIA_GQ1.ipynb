```python
"""
Academia de Herencia - Juego Educativo de Programación Orientada a Objetos (POO)

Autor: Camilo Rodríguez U00181176
Fecha: 2026

Propósito:
- Enseñar los conceptos de Herencia y Sobrescritura en POO de forma interactiva y divertida.
- Que el jugador aprenda a identificar cómo las habilidades se heredan y se agregan o modifican
  en distintas criaturas (analogía de clases y subclases).

Cómo jugar:
- Se presentarán 10 criaturas con habilidades base y habilidades especiales.
- Para cada criatura, debes elegir la opción correcta que combine todas sus habilidades.
- Cada nivel incluye explicación para reforzar el aprendizaje.

"""

import time
import random

class Nivel:
    def __init__(self, explicacion_previa, pregunta, opciones, correcta, explicacion_final):
        self.explicacion_previa = explicacion_previa
        self.pregunta = pregunta
        self.opciones = opciones
        self.correcta = correcta
        self.explicacion_final = explicacion_final
    def jugar(self):
        print("\n" + "="*60)
        print("CONTEXTO:")
        print(self.explicacion_previa)
        time.sleep(1)
        print("\nPREGUNTA:")
        print(self.pregunta)
        opciones_mezcladas = self.opciones[:]
        random.shuffle(opciones_mezcladas)
        texto_correcto = self.opciones[self.correcta - 1]
        nueva_correcta = opciones_mezcladas.index(texto_correcto) + 1

        for i, op in enumerate(opciones_mezcladas):
            print(f"{i+1}. {op}")

        while True:
            try:
                respuesta = int(input("Tu respuesta: "))
                break
            except:
                print("Ingresa un número válido")

        if respuesta == nueva_correcta:
            print("\nCorrecto")
            resultado = True
        else:
            print("\nIncorrecto")
            resultado = False

        print("\nEXPLICACIÓN:")
        print(self.explicacion_final)
        time.sleep(1)
        return resultado

def main():
    niveles = [
        Nivel(
            "El lobo hereda caminar y respirar, y añade su sonido: aullar.",
            "El lobo es una criatura. ¿Qué puede hacer?",
            ["Caminar solo", "Respirar solo", "Caminar y respirar", "Caminar, respirar y aullar"],
            4,
            "El lobo hereda caminar y respirar, y además aulla."
        ),
        Nivel(
            "El perro hereda caminar y respirar y añade su ladrido.",
            "El perro es una criatura. ¿Qué puede hacer?",
            ["Caminar", "Caminar y ladrar", "Caminar, respirar y ladrar", "Solo ladrar"],
            3,
            "El perro hereda caminar y respirar, y además ladra."
        ),
        Nivel(
            "El tigre hereda caminar y respirar y ataca con garras.",
            "El tigre es una criatura. ¿Qué puede hacer?",
            ["Caminar y respirar", "Caminar, respirar y garras", "Solo garras", "Caminar y garras"],
            2,
            "El tigre hereda lo base y añade sus garras como ataque."
        ),
        Nivel(
            "El dragón hereda caminar, respirar y atacar, y además puede lanzar fuego.",
            "El dragón es una criatura. ¿Qué puede hacer?",
            ["Solo lanzar fuego", "Caminar, respirar y fuego", "Caminar, respirar, atacar y fuego", "Solo atacar"],
            3,
            "El dragón mantiene lo base y añade fuego."
        ),
        Nivel(
            "El búho hereda caminar, respirar y hace 'hoo hoo'.",
            "El búho es una criatura nocturna. ¿Qué puede hacer?",
            ["Caminar, respirar y hoo hoo", "Solo sonido", "Caminar y respirar", "Solo hoo hoo"],
            1,
            "El búho hereda caminar y respirar, y además emite su sonido."
        ),
        Nivel(
            "El águila hereda caminar, respirar y puede volar.",
            "El águila es una criatura que vuela alto. ¿Qué puede hacer?",
            ["Volar", "Caminar y volar", "Caminar, respirar y volar", "Solo caminar"],
            3,
            "El águila hereda caminar y respirar y añade volar."
        ),
        Nivel(
            "El pez hereda nadar y respirar bajo el agua.",
            "El pez es una criatura acuática. ¿Qué puede hacer?",
            ["Nadar", "Caminar, respirar y nadar", "Solo respirar", "Caminar y nadar"],
            2,
            "El pez puede nadar y respirar, no caminar."
        ),
        Nivel(
            "El león hereda caminar, respirar y ataca, pero cambia ataque por rugido.",
            "El león es el rey de la selva. ¿Qué puede hacer?",
            ["Caminar, respirar y rugido", "Solo rugido", "Caminar y rugido", "Solo atacar"],
            1,
            "El león hereda caminar y respirar, y además ruge."
        ),
        Nivel(
            "El fénix hereda caminar, respirar y atacar, y añade renacer.",
            "El fénix es una criatura mítica. ¿Qué puede hacer?",
            ["Solo renacer", "Caminar, respirar, atacar y renacer", "Caminar y renacer", "Solo atacar"],
            2,
            "El fénix mantiene lo base y añade renacer."
        ),
        Nivel(
            "El dragón final hereda caminar, respirar y atacar, y añade fuego.",
            "¿Qué puede hacer el dragón?",
            ["Caminar, respirar, ataque y fuego", "Caminar, respirar y fuego", "Solo fuego", "Solo ataque"],
            1,
            "El dragón mantiene lo base y añade fuego."
        ),
    ]

    puntos = 0
    print("Academia de Herencia - Modo Comprensión\n")
    for nivel in niveles:
        if nivel.jugar():
            puntos += 1

    print("\n" + "="*60)
    print("Puntaje final:", puntos, "/", len(niveles))
    if puntos >= 8:
        print("Nivel alto de comprensión")
    elif puntos >= 5:
        print("Comprensión intermedia")
    else:
        print("Debes reforzar el concepto")

if __name__ == "__main__":
    main()
```
