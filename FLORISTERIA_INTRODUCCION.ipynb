```python
color_techo = input("Ingrese el color del techo: ")

color_pared_izquierda = input("Ingrese el color de la pared izquierda: ")
color_pared_derecha = input("Ingrese el color de la pared derecha: ")
color_pared_frontal = input("Ingrese el color de la pared frontal: ")
color_pared_trasera = input("Ingrese el color de la pared trasera: ")

color_piso = input("Ingrese el color del piso: ")

cantidad_puertas = int(input("Ingrese la cantidad de puertas: "))
cantidad_ventanas = int(input("Ingrese la cantidad de ventanas: "))
cantidad_baños = int(input("Ingrese la cantidad de baños: "))

color_marco_ventana = input("Ingrese el color del marco de la ventana: ")

puerta_abierta = input("¿La puerta está abierta? (si/no): ")

ventanas_abiertas = []

for i in range(cantidad_ventanas):
    estado = input(f"¿La ventana {i+1} está abierta? (si/no): ")
    if estado.lower() == "si":
        ventanas_abiertas.append(f"Ventana {i+1}")

decoracion_flores = input("¿Tiene decoración? (si/no): ")
flores = ""

if decoracion_flores.lower() == "si":
    flores = input("Ingrese la decoracion usada: ")

letrero = input("¿Tiene letrero? (si/no): ")
nombre_letrero = ""

if letrero.lower() == "si":
    nombre_letrero = input("Ingrese el nombre del letrero: ")

grafitis_pared = input("¿Tiene grafitis en las paredes? (si/no): ")
grafitis_techo = input("¿Tiene grafitis en el techo? (si/no): ")

diseno_flores_pared = input("¿Tiene diseños de flores en las paredes? (si/no): ")
diseno_flores_techo = input("¿Tiene diseños de flores en el techo? (si/no): ")

patio = input("¿Tiene patio? (si/no): ")
```

```python
def decorando_floristeria():

    todas_paredes_blancas = (
        color_pared_izquierda.lower() == "blanco" and
        color_pared_derecha.lower() == "blanco" and
        color_pared_frontal.lower() == "blanco" and
        color_pared_trasera.lower() == "blanco"
    )

    print("\nDESCRIPCIÓN DE LA FLORISTERÍA\n")

    print(f"Techo: {color_techo}")
    print(f"Pared izquierda: {color_pared_izquierda}")
    print(f"Pared derecha: {color_pared_derecha}")
    print(f"Pared frontal: {color_pared_frontal}")
    print(f"Pared trasera: {color_pared_trasera}")
    print(f"Piso: {color_piso}")

    print(f"\nPuertas: {cantidad_puertas}")
    print(f"Baños: {cantidad_baños}")
    print(f"Marco de ventana: {color_marco_ventana}")

    print("\nESTADO DEL LOCAL:")

    if puerta_abierta.lower() == "si":
        print("La puerta está abierta")
    else:
        print("La puerta está cerrada")

    if ventanas_abiertas:
        print("Ventanas abiertas:")
        for v in ventanas_abiertas:
            print(v)
    else:
        print("Todas las ventanas están cerradas")

    print("\nDECORACIÓN Y ARTE:")

    print(f"Grafitis en paredes: {grafitis_pared}")
    print(f"Grafitis en techo: {grafitis_techo}")

    print(f"Diseños de flores en paredes: {diseno_flores_pared}")
    print(f"Diseños de flores en techo: {diseno_flores_techo}")

    print(f"Patio: {patio}")

    if decoracion_flores.lower() == "si":
        print(f"Flores decorativas: {flores}")

    if letrero.lower() == "si":
        print(f"Letrero: {nombre_letrero}")

    if todas_paredes_blancas:
        print("\nTodas las paredes son blancas")

    if todas_paredes_blancas and diseno_flores_pared.lower() == "si":
        print("Paredes blancas con diseños de flores")

    if todas_paredes_blancas and grafitis_pared.lower() == "si":
        print("Paredes blancas con grafitis")
```
