##IF/ELSE
Evalua una o mas condicones, y dependiendo de su valor de verdad, realizara una accion u otra(solo dos acciones)

Sintaxis

if(condicon):
  Si la condicion es verdadera, se ejecuta este pedazo de codigo

else:
  Si la condicion es falsa, ejecuta este pedazo de codigo

```python
lluvia=input("Esta lloviendo?")
```

```python
if lluvia=="si" or lluvia== "sí" or lluvia=="si, mucho":
  print("Use sombrilla y guarde la ropa")
else:
  print("Salga tranquilo")
```

## IF TERNARIO
Sintaxis

variable=(codigo si condicion es verdadera) if(condicion) else(codigo si condicion es falsa)

```python
aguacero = "Use sombrilla y guarde la ropa" if lluvia=="si" else "Salga tranquilo"
```
