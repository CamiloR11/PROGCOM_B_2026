```python
class Persona:
    def __init__(self, nombre, genero, color_camisa, color_pantalon, zapato):
        self.nombre = nombre
        self.genero = genero
        self.color_camisa = color_camisa
        self.color_pantalon = color_pantalon
        self.zapato = zapato

    def posicion_inicial(self):
        print(f"{self.nombre}. posicion inicial")

    def levantar_pierna_derecha(self):
        print(f"{self.nombre}. levanta pierna derecha")

    def mantener_equilibrio(self):
        print(f"{self.nombre}. mantiene equilibrio")

    def bajar_pie_y_media_vuelta(self):
        print(f"{self.nombre}. baja el pie y da media vuelta")

    def levantar_pie_derecho_y_patear(self):
        print(f"{self.nombre}. levanta pie derecho y da patada")

    def dar_mini_saltos_laterales(self):
        print(f"{self.nombre}. da mini saltos hacia un lado")

    def levantar_pie_derecho_patadas_y_avanzar(self):
        print(f"{self.nombre}. levanta pie derecho, da patadas y avanza")

    def levantar_manos_y_cruzar_pie(self):
        print(f"{self.nombre}. levanta manos y cruza un pie detras del otro")

    def mover_piernas_dentro_fuera_en_saltos(self):
        print(f"{self.nombre}. mueve piernas hacia dentro y afuera avanzando en saltos")

    def estirar_mano_izquierda_y_pie_derecho(self):
        print(f"{self.nombre}. estira mano izquierda y pie derecho")

    def mover_caderas_lado_a_lado(self):
        print(f"{self.nombre}. mueve caderas de lado a lado")

    def hacerse_el_muerto(self):
        print(f"{self.nombre}. se hace el muerto")


# ── Crear bailarines ──────────────────────────────────────
camilo = Persona("Camilo", "M", "negra", "negro", "negro")
andres = Persona("Andres", "M", "beige", "jean azul", "ninguno")

# ── Baile en sincronía (los dos al mismo tiempo) ──────────
pasos = [
    "posicion_inicial",
    "levantar_pierna_derecha",
    "mantener_equilibrio",
    "bajar_pie_y_media_vuelta",
    "levantar_pie_derecho_y_patear",
    "dar_mini_saltos_laterales",
    "levantar_pie_derecho_patadas_y_avanzar",
    "levantar_manos_y_cruzar_pie",
    "dar_mini_saltos_laterales",
    "mover_piernas_dentro_fuera_en_saltos",
    "estirar_mano_izquierda_y_pie_derecho",
    "dar_mini_saltos_laterales",
    "mover_caderas_lado_a_lado",
    "hacerse_el_muerto",
]

for paso in pasos:
    getattr(camilo, paso)()
    getattr(andres, paso)()
```
