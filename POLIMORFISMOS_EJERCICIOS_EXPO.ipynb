```python
class Animal:
    def sonido(self):
        print("El animal hace un sonido")

class Perro(Animal):
    def sonido(self):
        print("Guau guau")

class Gato(Animal):
    def sonido(self):
        print("Miau")
```

```python
perro = Perro()
gato = Gato()

perro.sonido()
gato.sonido()
```

```python
class Calculadora:

    def operar(self, a, b, operador="+"):

        if operador == "+":
            print(a + b)

        elif operador == "-":
            print(a - b)
```

```python
calc = Calculadora()

calc.operar(10, 5)
calc.operar(10, 5, "-")
```

```python
#1
class Animal:

    def sonido(self):
        print("El animal hace un sonido")


class Perro(Animal):

    def sonido(self):
        print("Guau guau")


class Gato(Animal):

    def sonido(self):
        print("Miau")


class Vaca(Animal):

    def sonido(self):
        print("Muuu")


perro = Perro()
gato = Gato()
vaca = Vaca()

perro.sonido()
gato.sonido()
vaca.sonido()
```

```python
#2
class Vehiculo:

    def mover(self):
        print("El vehículo se mueve")


class Carro(Vehiculo):

    def mover(self):
        print("El carro se mueve por carretera")


class Avion(Vehiculo):

    def mover(self):
        print("El avión vuela por el aire")


class Barco(Vehiculo):

    def mover(self):
        print("El barco navega por el agua")


carro = Carro()
avion = Avion()
barco = Barco()

carro.mover()
avion.mover()
barco.mover()
```

```python
#3
class Empleado:

    def trabajar(self):
        print("El empleado trabaja")


class Profesor(Empleado):

    def trabajar(self):
        print("El profesor enseña estudiantes")


class Ingeniero(Empleado):

    def trabajar(self):
        print("El ingeniero diseña proyectos")


class Doctor(Empleado):

    def trabajar(self):
        print("El doctor atiende pacientes")


profesor = Profesor()
ingeniero = Ingeniero()
doctor = Doctor()

profesor.trabajar()
ingeniero.trabajar()
doctor.trabajar()
```

```python
#4
class Dispositivo:

    def encender(self):
        print("El dispositivo se enciende")


class Televisor(Dispositivo):

    def encender(self):
        print("El televisor muestra imagen")


class Celular(Dispositivo):

    def encender(self):
        print("El celular inicia aplicaciones")


class Computador(Dispositivo):

    def encender(self):
        print("El computador carga el sistema")


tv = Televisor()
cel = Celular()
pc = Computador()

tv.encender()
cel.encender()
pc.encender()
```

```python
#5
class Deporte:

    def jugar(self):
        print("Se practica un deporte")


class Futbol(Deporte):

    def jugar(self):
        print("El fútbol se juega con un balón")


class Baloncesto(Deporte):

    def jugar(self):
        print("El baloncesto usa una canasta")


class Tenis(Deporte):

    def jugar(self):
        print("El tenis utiliza raquetas")


futbol = Futbol()
basket = Baloncesto()
tenis = Tenis()

futbol.jugar()
basket.jugar()
tenis.jugar()
```

```python
#6
class Instrumento:

    def tocar(self):
        print("El instrumento produce sonido")


class Guitarra(Instrumento):

    def tocar(self):
        print("La guitarra toca acordes")


class Piano(Instrumento):

    def tocar(self):
        print("El piano produce melodías")


class Bateria(Instrumento):

    def tocar(self):
        print("La batería marca el ritmo")


guitarra = Guitarra()
piano = Piano()
bateria = Bateria()

guitarra.tocar()
piano.tocar()
bateria.tocar()
```

```python
#7
class Transporte:

    def viajar(self):
        print("El transporte realiza un viaje")


class Bus(Transporte):

    def viajar(self):
        print("El bus transporta pasajeros")


class Moto(Transporte):

    def viajar(self):
        print("La moto se desplaza rápidamente")


class Bicicleta(Transporte):

    def viajar(self):
        print("La bicicleta avanza con pedales")


bus = Bus()
moto = Moto()
bici = Bicicleta()

bus.viajar()
moto.viajar()
bici.viajar()
```

```python
#8
class Personaje:

    def atacar(self):
        print("El personaje realiza un ataque")


class Mago(Personaje):

    def atacar(self):
        print("El mago lanza un hechizo")


class Arquero(Personaje):

    def atacar(self):
        print("El arquero dispara una flecha")


class Guerrero(Personaje):

    def atacar(self):
        print("El guerrero usa una espada")


mago = Mago()
arquero = Arquero()
guerrero = Guerrero()

mago.atacar()
arquero.atacar()
guerrero.atacar()
```

```python
#9
class Calculadora:

    def operar(self, a, b, operador="+"):

        if operador == "+":
            print(a + b)

        elif operador == "-":
            print(a - b)


calc = Calculadora()

calc.operar(10, 5)
calc.operar(10, 5, "-")
```

```python
#10
class Texto:

    def convertir(self, mensaje, tipo="mayus"):

        if tipo == "mayus":
            print(mensaje.upper())

        elif tipo == "minus":
            print(mensaje.lower())


texto = Texto()

texto.convertir("Hola Mundo")
texto.convertir("Hola Mundo", "minus")
```
